```python
cols = ["temperature", "humidity", "ph", "rainfall"]
df1[cols] = df1[cols].round(2)
```

```python
target = "label"

labels1 = set(df1[target])
labels2 = set(df2[target])

print("Crops only in Dataset 1:", labels1 - labels2)
print("Crops only in Dataset 2:", labels2 - labels1)
print("Common crops:", labels1 & labels2)
```

```python
print("\nExact duplicate rows in Dataset 1:", df1.duplicated().sum())
print("Exact duplicate rows in Dataset 2:", df2.duplicated().sum())


# remove duplicates from df2
df2 = df2.drop_duplicates().reset_index(drop=True)
```

```python
# print("Dataset 1 statistics:")
# print(df1.describe().round(2))

# print("\nDataset 2 statistics:")
# print(df2.describe().round(2))


print(
    df1.groupby("label")[["N", "P", "K", "temperature", "humidity", "ph", "rainfall"]]
    .mean()
    .round(2)
)

print(
    df2.groupby("label")[["N", "P", "K", "temperature", "humidity", "ph", "rainfall"]]
    .mean()
    .round(2)
)
```

```python
features = [
    "N", "P", "K", "temperature",
    "humidity", "ph", "rainfall"
]

stats1 = df1.groupby("label")[features].mean().round(2)
stats2 = df2.groupby("label")[features].mean().round(2)

matches = []

for label1, row1 in stats1.iterrows():
    for label2, row2 in stats2.iterrows():
        same = row1.eq(row2)

        if same.all():
            matches.append({
                "Dataset 1 label": label1,
                "Dataset 2 label": label2,
                "Matching features": len(features)
            })

matches_df = pd.DataFrame(matches)
print(matches_df)
```

```python
features = [
    "N", "P", "K", "temperature",
    "humidity", "ph", "rainfall"
]

pairs = [
    ("blackgram", "cassava"),
    ("kidneybeans", "beans"),
    ("mungbean", "yam"),
    ("muskmelon", "tomatoes"),
    ("pomegranate", "guinea_corn"),
    ("watermelon", "pepper")
]

for crop1, crop2 in pairs:
    a = df1.loc[df1["label"] == crop1, features].reset_index(drop=True)
    b = df2.loc[df2["label"] == crop2, features].reset_index(drop=True)

    matches = a.merge(b.drop_duplicates(), on=features)

    print(f"{crop1} vs {crop2}: {len(matches)} matching feature rows")
```

### Last Work Combining Both Dataset

```python
#Check duplicates before combining datasets
features = [
    "N", "P", "K", "temperature",
    "humidity", "ph", "rainfall", "label"
]

df1 = df1[features].copy()
df2 = df2[features].copy()

df2 = df2.drop_duplicates()

combined = pd.concat([df1, df2], ignore_index=True)

combined = combined.drop_duplicates().reset_index(drop=True)

print("Combined rows:", len(combined))
print("Combined exact duplicates:", combined.duplicated().sum())
```

```python
print(combined["label"].value_counts())
print(combined["label"].unique())
```

```python
combined.shape
```

```python
features = [
    "N", "P", "K", "temperature",
    "humidity", "ph", "rainfall"
]

conflicts = (
    combined.groupby(features)["label"]
      .agg(["nunique", "unique", "size"])
      .query("nunique > 1")
      .sort_values("size", ascending=False)
)

print(conflicts)
```

```python
# 1. Count conflicting rows and identify affected crops
features = [
    "N", "P", "K", "temperature",
    "humidity", "ph", "rainfall"
]

# Feature combinations associated with multiple labels
conflicts = (
    combined.groupby(features)["label"]
      .agg(["nunique", "unique", "size"])
      .query("nunique > 1")
)

print("Conflicting feature combinations:", len(conflicts))
print("Rows involved in conflicts:", conflicts["size"].sum())

# Count conflicting rows by crop label
conflict_rows = combined.merge(
    conflicts.reset_index()[features],
    on=features,
    how="inner"
)

print("\nConflicting rows by crop:")
print(conflict_rows["label"].value_counts())
```

```python
# Save combined dataset
combined.to_csv("../dataset/processed/merged_dataset.csv", index=False)
```
